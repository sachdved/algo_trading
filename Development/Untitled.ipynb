```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
gamma = np.linspace(0.001, 1000, 1000000)
```

```python
plt.plot(gamma, np.abs(np.log(gamma)), label = 'Log error')
plt.plot(gamma, np.abs(gamma - 1), label = 'MAPE')
plt.plot(gamma, np.abs(2 * (gamma-1)/(gamma+1)), label = 'SMAPE')

plt.legend()
plt.xlabel('$\hat{Y}/Y$')
plt.ylabel('Loss fn')
plt.xscale('log')
plt.yscale('log')
```
